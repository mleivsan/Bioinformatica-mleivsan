##### 4.- Escribe una función que cuente el porcentaje de A de una secuencia

```python
def porcentaje_A(seq):
    if not seq:
        return 0.0
    seq = seq.upper()
    return (seq.count('A') / len(seq)) * 100

print(porcentaje_A("AaATGCa"))
```

##### ¿Puedes hacer que devuelva dos decimales? ¿Y que lidie con mayúsculas y minúsculas?

```python
def porcentaje_A(seq):
    if not seq:
        return 0.0
    seq = seq.upper()
    return round((seq.count('A') / len(seq)) * 100, 2)

print(porcentaje_A("AaATGCa"))
```

##### ¿Y que devuelva el número de decimales que yo quiero?

```python
def porcentaje_A(seq, decimales):
    if not seq:
        return 0.0
    seq = seq.upper()
    return round((seq.count('A') / len(seq)) * 100, decimales)

print(porcentaje_A("AaATGCa", 3))
```

##### ¿Sabes poner un valor por defecto en un argumento?

```python
def porcentaje_A(seq, decimales=2):
    if not seq:
        return 0.0
    seq = seq.upper()
    return round((seq.count('A') / len(seq)) * 100, decimales)

print(porcentaje_A("AaATGCa"))
print(porcentaje_A("AaATGCa", 4))
```

##### ¿Qué pasa si no ponemos un argumento? ¿Sabrías hacer una función encapsulada?

```python
#Si no se pone 'seq', la función da error TypeError porque falta el argumento obligatorio
def porcentaje_A_encapsulada(seq, decimales=2):
    def limpiar(s):
        return s.upper().strip()
    if not seq:
        return 0.0
    s = limpiar(seq)
    return round((s.count('A') / len(s)) * 100, decimales)

print(porcentaje_A_encapsulada("AaATGCa"))
```

##### Testea tu función original con assert

```python
def porcentaje_A(seq, decimales=2):
    if not seq:
        return 0.0
    seq = seq.upper()
    return round((seq.count('A') / len(seq)) * 100, decimales)

assert porcentaje_A("AAAA") == 100.00
assert porcentaje_A("TTTT") == 0.00
assert porcentaje_A("AaTt") == 50.00
assert porcentaje_A("") == 0.00

print("AaATGCa")
```
